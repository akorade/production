```python
import dask.dataframe as dd
```

+ Load the environment variable `PRICE_DATA`.
+ Use [glob](https://docs.python.org/3/library/glob.html) to find the path of all parquet files in the directory `PRICE_DATA`.

(1pt)

```python
import os
from glob import glob

# Write your code below.
price_data = os.getenv('PRICE_DATA')
parquet_files = glob(os.path.join(price_data, '**/*.parquet'), recursive=True)
print(f"Found {len(parquet_files)} parquet files")
print(f"Sample files: {parquet_files[:3]}")
```

For each ticker and using Dask, do the following:

+ Add lags for variables Close and Adj_Close.
+ Add returns based on Close:
    
    - `returns`: (Close / Close_lag_1) - 1

+ Add the following range: 

    - `hi_lo_range`: this is the day's High minus Low.

+ Assign the result to `dd_feat`.

(4 pt)

```python
# Write your code below.
dd_feat = None

for file in parquet_files:
    # Read parquet file with Dask
    df = dd.read_parquet(file)
    
    # Add lags for Close and Adj_Close
    df['Close_lag_1'] = df['Close'].shift(1)
    df['Adj_Close_lag_1'] = df['Adj Close'].shift(1)
    
    # Add returns based on Close
    df['returns'] = (df['Close'] / df['Close_lag_1']) - 1
    
    # Add hi_lo_range
    df['hi_lo_range'] = df['High'] - df['Low']
    
    # Combine all features
    if dd_feat is None:
        dd_feat = df
    else:
        dd_feat = dd.concat([dd_feat, df])

# Reset index to avoid conflicts
dd_feat = dd_feat.reset_index(drop=True)
```

+ Convert the Dask data frame to a pandas data frame. 
+ Add a new feature containing the moving average of `returns` using a window of 10 days. There are several ways to solve this task, a simple one uses `.rolling(10).mean()`.

(3 pt)

```python
# Write your code below.
# Convert Dask dataframe to pandas
df_feat = dd_feat.compute()

# Add moving average of returns with 10-day window
# Sort by Ticker and Date to ensure proper ordering
df_feat = df_feat.sort_values(['Ticker', 'Date'])
df_feat['returns_ma_10'] = df_feat.groupby('Ticker')['returns'].rolling(10).mean().reset_index(level=0, drop=True)
```
